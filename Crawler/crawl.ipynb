```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.action_chains import ActionChains
from time import sleep
```

```python
# Set up Chrome options
chrome_options = Options()
chrome_options.add_argument("--start-fullscreen")
browser = webdriver.Chrome(options=chrome_options)
browser.get('https://vietstock.vn/chung-khoan.htm')
url_csv = []
for i in range(1, 200):
    sleep(2)
    a_fontbold = browser.find_elements(By.CSS_SELECTOR, 'a.fontbold')
    hrefs = [element.get_attribute("href") for element in a_fontbold]
    for href in hrefs:
        url_csv.append(href)
    next_button = browser.find_element(By.CSS_SELECTOR, 'a[title="Trang sau"]')
    actions = ActionChains(browser)
    actions.move_to_element(next_button).perform()
    next_button.click()

browser.quit()
```

```python
url_csv = list(set(url_csv))
url_csv = pd.DataFrame(url_csv, columns=['url'])
url_csv.to_csv('url.csv', index=False)
```

```python
browser = webdriver.Chrome()
news_data = []
for url in url_csv['url']:
    browser.get(url)
    try:
        title = browser.find_element(By.CSS_SELECTOR, 'h1[class="article-title"]').text
        date = browser.find_element(By.CSS_SELECTOR, 'span[class="date"]').text
        content = browser.find_elements(By.CSS_SELECTOR, 'p[class="pBody"]')
        content = [element.text for element in content]
        content = ' '.join(content)
        link = url
        news_data.append([title, link, content, date])
    except:
        pass
```

```python
news_data = pd.DataFrame(news_data, columns=['title', 'link', 'content', 'date'])
news_data.to_csv('news_data.csv', index=False)
```
